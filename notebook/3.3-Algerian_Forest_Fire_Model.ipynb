### Algerian Forst Fire Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.head()
```

```python
df.info()
```

### Data Cleaning

```python
### Missing Values
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,'Region']=0
df.loc[122:,'Region']=1
```

```python
df.head()
```

```python
df.info()
```

```python
df['Region']=df['Region'].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
## Remove the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## Fix The spaces In column names
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
### Change The Required Columns to int

df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
## Changing The Other Columns to float\
objects=[features for features in df.columns if df[features].dtype=='str']
```

```python
objects
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
## Lets save the Cleaned Dataset
df.to_csv('Algerian_Forest_Fires_Cleaned_Dataset.csv',index=False)
```

## EDA

```python
### Drop day,month,year feature
df_copy=df.copy()
```

```python
df.head()
```

```python
df_copy.drop(['day','month','year'],axis=1,inplace=True)
```

```python
df_copy.head()
```

```python
## Categories In Classes
df_copy['Classes'].value_counts()
```

```python
## Encoding of the Categories in Classes
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
df_copy.tail()
```

```python
df_copy['Classes'].value_counts()
```

```python
## Plot density plot for all features

plt.figure(figsize=(15, 20))
plt.style.use('seaborn-v0_8')
for i,column in enumerate(df_copy.columns):
    plt.subplot(6,2,i+1)
    sns.histplot(df_copy[column],bins=50) 
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python
## OR The Alternate
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
plt.pie(df_copy['Classes'].value_counts(),labels=['Not Fire','Fire'],autopct='%1.1f%%')
```

```python
## Percentage for pie chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
percentage
```

```python
### Plotting Pie Chart
classlabels=['Fire','Not Fire']
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.show() 
```

```python
## Monthly Fire Analysis
dftemp=df.loc[df['Region']==0]
plt.figure(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=dftemp)
plt.xlabel('Month',weight='bold')
plt.ylabel('Number of Fire',weight='bold')
plt.title('Fire Analysis of Sidi- Bel',weight='bold')
plt.show()
```
